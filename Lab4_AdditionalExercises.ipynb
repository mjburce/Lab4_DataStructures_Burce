```python
SEED_NUM = 8
STUDENT_MAJOR = "BSME"
FAVORITE_ARTIST = "fitterkarma"
LOCAL_HAZARD = "TYPHOON"
CONTROL_NUM = max(3, SEED_NUM)
```

EXERCISE 1

```python
from collections import defaultdict, Counter

stream = [
    ("TUP Manila", "BSECE"),
    ("TUP Taguig", f"BSME_v{CONTROL_NUM}"),
    ("TUP Manila", f"BSME_v{CONTROL_NUM}"),
    ("TUP Manila", "BSECE"),
    ("TUP Visayas", STUDENT_MAJOR),
    ("TUP Taguig", "BSECE"),
    ("TUP Manila", "BSECE")
]

custom_data = [
    ("TUP Manila", "BSME"), ("TUP Manila", "BSME"), ("TUP Manila", "BSME"),
    ("TUP Taguig", "BSIT"), ("TUP Taguig", "BSIT"), ("TUP Visayas", "BSCE"),
    ("TUP Manila", "BSME"), ("TUP Manila", "BSME"), ("TUP Manila", "BSME"),
    ("TUP Taguig", "BSME"), ("TUP Manila", "BSME")
]
stream.extend(custom_data)

campus_groups = defaultdict(list)
for campus, program in stream:
    campus_groups[campus].append(program)

manila_counter = Counter(campus_groups["TUP Manila"])
top_program_manila, frequency = manila_counter.most_common(1)[0]

print("--- EXERCISE 1 OUTPUT ---")
print(f"CONTROL_NUM Used: {CONTROL_NUM}")
print(f"STUDENT_MAJOR Used: {STUDENT_MAJOR}")
print(f"Expanded Stream Length: {len(stream)}")
print(f"Total Applications Processed for TUP Manila: {len(campus_groups['TUP Manila'])}")
print(f"Top Requested Program at TUP Manila: {top_program_manila}")
print(f"Exact Frequency of Top Program: {frequency}")
```

```python
festival = {"Ben&Ben", "SB19", "Bini", "Eraserheads", FAVORITE_ARTIST, "Zild", f"Indie Artist {CONTROL_NUM}"}
user_a = {"Ben&Ben", "Bini", "Maki", "Dionela", FAVORITE_ARTIST}
user_b = {"SB19", "Eraserheads", "Zild", f"Indie Artist {CONTROL_NUM}", "Parokya ni Edgar"}

new_artists = {"Lola Amour", "Adie", "Cup of Joe", "Dilaw", "Munimuni", "Juan Karlos", "IV of Spades", "fitterkarma"}
festival.update(new_artists)

intersection_a = user_a.intersection(festival)
union_a = user_a.union(festival)
jaccard_a = (len(intersection_a) / len(union_a)) * 100

intersection_b = user_b.intersection(festival)
union_b = user_b.union(festival)
jaccard_b = (len(intersection_b) / len(union_b)) * 100

dealbreakers_a = user_a.difference(festival)

print("--- EXERCISE 2 OUTPUT ---")
print(f"Total Artists in Expanded Festival: {len(festival)}")
print(f"Length of Intersection (User A & Festival): {len(intersection_a)}")
print(f"Length of Union (User A & Festival): {len(union_a)}")
print(f"User A Jaccard Similarity (%): {jaccard_a:.2f}%")
print(f"User B Jaccard Similarity (%): {jaccard_b:.2f}%")
print(f"User A Dealbreaker Artists (Set Output): {dealbreakers_a}")
```

```python
from collections import deque, Counter

buffer = deque(maxlen=5)

burst = [
    (CONTROL_NUM, "WEATHER"),
    (CONTROL_NUM + 1, "TRAFFIC"),
    (CONTROL_NUM + 2, "WEATHER"),
    (CONTROL_NUM + 3, "WEATHER"),
    (CONTROL_NUM + 4, LOCAL_HAZARD),
    (CONTROL_NUM + 5, "WEATHER"),
    (CONTROL_NUM + 6, "TRAFFIC")
]

for i in range(10):
    timestamp = CONTROL_NUM + 7 + i
    category = "FIRE" if i % 2 == 0 else "TRAFFIC"
    burst.append((timestamp, category))

weather_3_append = "N"
hazard_append = "N"

for alert in burst:
    current_categories = [item[1] for item in buffer]
    counts = Counter(current_categories)
    
    incoming_category = alert[1]
    
    if counts[incoming_category] < 2:
        buffer.append(alert)
        # Tracking for the assessment table
        if alert[0] == CONTROL_NUM + 3: weather_3_append = "Y"
        if alert[1] == LOCAL_HAZARD: hazard_append = "Y"

timestamp_sum = sum(alert[0] for alert in buffer)

print("--- EXERCISE 3 OUTPUT ---")
print(f"LOCAL_HAZARD Used: {LOCAL_HAZARD}")
print(f"Total Burst Size: {len(burst)}")
print(f"Did 'WEATHER' at Timestamp {CONTROL_NUM + 3} append? {weather_3_append}")
print(f"Did {LOCAL_HAZARD} append? {hazard_append}")
print(f"Final Deque Length: {len(buffer)}")
print(f"Final Sum of Timestamps in Deque: {timestamp_sum}")
print(f"Exact Final Deque Output: {list(buffer)}")
```
